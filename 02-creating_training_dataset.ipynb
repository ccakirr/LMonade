## Creating Training Dataset

```python
from tokenizers import Tokenizer
from dataset import GPTDataset
from torch.utils.data import DataLoader

import datasets
import torch
```

```python
# loading data

dataset = datasets.load_dataset(
    "wikimedia/wikipedia",
    "20231101.tr",
    split="train",
)

texts = [article["text"] for article in dataset.take(534988)]
```

```python
# loading tokenizer

tokenizer = Tokenizer.from_file("turkish_tokenizer.json")
```

```python
# creating dataset

ids = []

for text in texts:
    ids.extend(tokenizer.encode(text).ids)

dataset = GPTDataset(ids, context_length=1024)

dataloader = DataLoader(
    dataset,
    batch_size=4
)
```

```python
# checking dataset

x, y = next(iter(dataloader))

print(x.shape)
print(y.shape)
```

```python
# saving dataset

torch.save(dataset.ids, "data/turkish_dataset.pt")
```
